```python
import re
import math
from collections import Counter
```

```python
class TFIDFTokenizer:
    def __init__(self):
        self.documents_count: int = None
        self.idf_: dict[str: float] = None
        self.corpus_: list[str] = None

    def fit(self, documents: list[str]):
        docs = [self.preprocess_text(document) for document in documents]

        self.documents_count = len(docs)
        self.idf_ = self._compute_idf(docs)
        self.corpus_ = self.idf_.keys()

    def transform(self, document: str) -> dict:
        document_tokens = self.preprocess_text(document)
        return self.compute_tf_idf(document_tokens)

    def fit_transform(self, documents: list[str]) -> list[dict]:
        self.fit(documents)
        return [self.transform(document) for document in documents]

    def preprocess_text(self, text: str) -> list:
        # case Normalization
        text = text.lower()

        # delete puncetuation
        re.sub(r'[^a-zA-Z0-9\s-]', ' ', text)
        re.sub(r'\s+', ' ', text)

        # tokenization
        return text.split()

    def _compute_tf(self, document_tokens: list) -> dict:
      total_tokens = len(document_tokens)
      counts = Counter(document_tokens)

      tf = {
            token: count / total_tokens
            for token, count in counts.items()
          }

      return tf

    def _compute_idf(self, documents: list[list[str]]) -> dict:
        tokens = []
        for i in range(self.documents_count):
            tokens += list(set(documents[i]))

        token_counts = Counter(tokens)

        idf = {
            token: math.log(self.documents_count / count + 1) + 1
            for token, count in token_counts.items()
        }

        return idf


    def compute_tf_idf(self, document_tokens: list) -> dict:

      tf = self._compute_tf(document_tokens)
      non_existed_tokens_idf = math.log(self.documents_count) + 1

      return {
            token: tf[token] * self.idf_.get(token, non_existed_tokens_idf)
            for token in tf.keys()
      }

      return tf_idf


```
